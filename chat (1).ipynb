```python
import string
import pickle
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

# NLTK data download (pehli baar run karne ke liye)
nltk.download('stopwords')
nltk.download('punkt')

# 1. Custom text cleaning function define karein
ps = PorterStemmer()
stop_words = set(stopwords.words('english'))

def clean_text(text):
    text = str(text).lower()
    tokens = word_tokenize(text)
    tokens = [ps.stem(word) for word in tokens if word.isalnum() and word not in stop_words]
    return " ".join(tokens)

# 2. Text clean karein
df["clean_text"] = df["text"].apply(clean_text)

# 3. Vectorizer & Model training
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(df["clean_text"])
y = df["intent"]  # Apne dataset ke target column name ke hisaab se change kar lein

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)

# 4. Save model & vectorizer
pickle.dump(model, open("chatbot_model.pkl", "wb"))
pickle.dump(vectorizer, open("vectorizer.pkl", "wb"))

print("Sab kuch successful run aur save ho gaya!")
```

```python
def predict_intent(text):
    cleaned_text = clean_text(text)
    vector = vectorizer.transform([cleaned_text])
    prediction = model.predict(vector)
    return prediction[0]
```

```python
while True:
    text = input("Enter your message (type 'exit' to quit): ")

    if text.lower() == "exit":
        break

    print("Predicted Intent:", predict_intent(text))
```
